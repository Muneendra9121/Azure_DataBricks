```python
#StructType,StructField(for nested struct,map,array)
#StringType()
#IntegerType()
#FloatType()
#DateType()
#TimeStampe()
'''Function	Description
current_date()	Gets current system date
current_timestamp()	Gets current date & time
to_date()	Converts string to DateType
to_timestamp()	Converts string to TimestampType
datediff()	Days between two dates
date_add()	Adds days to date
date_sub()	Subtracts days from date
year(), month(), dayofmonth()'''

from pyspark.sql.types import DateType
from pyspark.sql import Row
from datetime import date
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,ArrayType,MapType
d=[(1,'ganesh',2000),(2,'maha',40000)]
schema=StructType([StructField(name='id',dataType=IntegerType(),nullable=False),\
     StructField(name='name',dataType=StringType(),nullable=False),\
          StructField(name='salary',dataType=IntegerType(),nullable=False)])

df=spark.createDataFrame(data=d,schema=schema)
df.show()
df.printSchema() 

```

```python
#how to read struct type data column like((1,(3,7))
from pyspark.sql.types import StructType,StructField,IntegerType,StringType
from pyspark.sql.functions import col
data=[(1,(20,'maha')),(2,(30,'ganesh'))]
proper=StructType([StructField('age',IntegerType()),StructField('name',StringType())])
schema=StructType([StructField('id',IntegerType()),StructField('addre',proper)])

df=spark.createDataFrame(data=data,schema=schema)
#df1=df.select(df.id,df.addre.age)
df2=df.select(df['addre.name'])
df3=df.select(col('addre.name'))
df3.show()
df2.show()
```
